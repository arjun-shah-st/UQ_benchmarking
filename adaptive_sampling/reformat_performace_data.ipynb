```python
import os 
import json
import numpy as np

filepath = os.path.join("CL_5D copy", "US_GP", "performance_data copy.json")

def reformat_data(filepath):
    with open(filepath, 'r') as f: 
        data = json.load(f)
        dicts = [eval(step) for step in data]
        dict_to_save = {} 
        for i, perf in enumerate(dicts): 
                if i==0: 
                    for key, value in perf.items(): 
                        dict_to_save[key] = [value]
                else: 
                    for key, value in perf.items(): 
                        dict_to_save[key].append(value)

        with open(os.path.join(filepath), 'w') as f: 
            json.dump(dict_to_save, f, indent=4)

methods = ["CQR_GRUBS", "GRUBS", "PREV", "RAND", "US_GP"]

for method in methods: 
     reformat_data(os.path.join("CM_5D", method, "performance_data.json"))
```
